1. Prevalence & incidence
Alzheimer’s disease affects approximately 7.4 million older adults, or about one in nine people aged 65 and older (https://www.alz.org/alzheimers-dementia/facts-figures). The risk of developing Alzheimer’s increases with age, from roughly 0.5% among people aged 65–70 to about 5–8% among those over 85 (https://pmc.ncbi.nlm.nih.gov/articles/PMC3405821/).

2. Economic burden
Alzheimer’s treatment can cost $468.28 annually for mild cases and up to $117,000 for severe cases, with costs expected to reach $2 trillion in the US by 2030 (https://pubmed.ncbi.nlm.nih.gov/37972428/).

3. Risk factors (genetic, lifestyle)
Genetic risk factor: ApoE genotype. One allele from each parent: “allele 1 and allele 2”. Alleles can be one of the three types: ε2 (protective), ε3 (neutral), or ε4 (most risk) (class slides)

4. Societal determinants
Lower levels of formal education may increase the risk of dementia because they are associated with lower cognitive reserve. Studies have also found that higher dementia can be caused by neighborhood disorder, social vulnerability, social isolation, and weaker community connections. Environmental factors, including geography and pollution, income, employment, working conditions, socioeconomic status, and poverty may play a role (https://pmc.ncbi.nlm.nih.gov/articles/PMC12301702/).

5. Symptoms
The most common symptoms of onset Alzheimer's is memory loss that disrupts daily life, difficulty completing daily tasks, and challenges in planning or solving problems (https://www.cdc.gov/alzheimers-dementia/signs-symptoms/alzheimers.html).

6. Diagnosis
- Diagnosis includes many steps:
    - Review of medical and family history 
    - Physical and neurological exams: tests reflexes, muscle tone and strength, sight and hearing, coordination, balance, ability to get up from a chair and walk across the room 
- Lab test 
    - Blood tests can help rule out other potential causes of memory loss, confusion, and other symptoms
    - Newer blood tests are able to measure levels of beta-amyloid protein and tau protein 
- Mental status and neuropsychological testing
    - Just a status check to check memory and thinking skills
- Brain imaging
    - Magnetic resonance imaging (MRI): use radio waves and a strong magnetic field to produce detailed images of the brain 
    - Computerized tomography (CT): specialized X-ray that produces cross sectional images of the brain 
    - PET scan: captures images of the disease process 
        - Fluorodeoxyglucose, also called FDG, PET imaging
        - Amyloid PET imaging
        - Tau PET imaging  (https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/diagnosis-treatment/drc-20350453)

7. Standard of care treatments (& reimbursement) (Google AI)
- Treatment include removing the beta amyloid plaque 
- Medicare is the usual target for reimbursements
- Medicare part D covers 80% of the medical expenses
- Lequimbi is a treatment used as well and it is IV infused every two weeks
- Another medication is Kisunla which is IV infused every month

8. Disease progression & prognosis
- Preclinical Phase: when the brain changes through the buildup of proteins (10 to 20 yrs before any outward signs appear)
- Mild Stage: when a person begins to feel mild cognitive impairment (misplacing items, forgetting familiar words, etc)
- Moderate Stage: confusion deepens, memory loss grows, personality changes, people may wander off. Require help with daily tasks such as dressing and bathing (longest stage)
- Severe Stage: physical and mental abilities decline (losing ability to speak, walk, swallow, and control their bodily functions), which requires care (https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/in-depth/alzheimers-stages/art-20048448).

9. Continuum of care providers
- Medical care needs to be continuous for doctors appointments to track the progression of the Alzheimer's and to monitor treatment
- Once Alzheimer's worsens at home care is also needed as people with Alzheimer's are no longer able to take care of themselves
- Expenses also go towards continued care for memory retention
- Patients also need to get end of life care as continued care (https://pmc.ncbi.nlm.nih.gov/articles/PMC6925963/)

10. Biological mechanisms (anatomy, organ physiology, cell & molecular physiology) 
Alzheimer’s affects the hippocampus, the entorhinal cortex, and other connected structures involved in memory. In Alzheimer’s disease, amyloid-beta (Aβ) plaques form outside neurons, while neurofibrillary tangles made of abnormal tau form inside them. There is also a loss of neurons and synapses which results in a cognitive decline. (https://pmc.ncbi.nlm.nih.gov/articles/PMC5290713/)

11. Clinical Trials/next-gen therapies (Google AI)
There are currently 158 new drugs across approximately 192 active clinical trials 
Phase 1: has 45 drugs in 49 trials testing basic safety and dosage 
Phase 2: has 84 drugs in 89 trials testing effectiveness and side effects
Phase 3: has 36 drugs in 54 large scale trials for confirming benefits before public approval 
Next-gen therapies: 
Trontinemab: an advanced antibody that crosses the brain’s protective shield faster to clear plaques 
Oral pills: scientists are creating daily pills to replace time consuming clinical IV infusions 
Combination drugs: new trials are mixing medications in order to attack two toxic brain proteins at once
