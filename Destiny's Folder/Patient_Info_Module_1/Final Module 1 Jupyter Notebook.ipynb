## Disease Background: 

1. Prevalence & incidence
- Alzheimer’s disease affects approximately 7.4 million older adults, or about one in nine people aged 65 and older (https://www.alz.org/alzheimers-dementia/facts-figures). The risk of developing Alzheimer’s increases with age, from roughly 0.5% among people aged 65–70 to about 5–8% among those over 85 (https://pmc.ncbi.nlm.nih.gov/articles/PMC3405821/).

2. Economic burden
- Alzheimer’s treatment can cost $468.28 annually for mild cases and up to $117,000 for severe cases, with costs expected to reach $2 trillion in the US by 2030 (https://pubmed.ncbi.nlm.nih.gov/37972428/).

3. Risk factors (genetic, lifestyle)
- Some major risk factors of Alzheimer's disease include age, family history, genetics, head injuries, or medical conditions. After the age of 65, the risk of developing Alzheimer's doubles and after the age of 85, there is a ⅓ risk. Additionally, if there is a history of a parent or sibling having Alzheimer's, an individual is more likely to develop it. Generics also contribute to Alzheimer's disease. There are two categories of genes including risk genes and deterministic genes. Risk genes increase the likelihood of developing Alzheimer’s, while deterministic genes directly cause it. However, deterministic genes account for less than 1% of Alzheimer’s cases. Furthermore, head injuries and certain medical conditions related to the heart, blood pressure, and diabetes can also be a risk of dementia (https://www.alz.org/alzheimers-dementia/what-is-alzheimers/causes-and-risk-factors)

4. Societal determinants
- Lower levels of formal education may increase the risk of dementia because they are associated with lower cognitive reserve. Studies have also found that higher dementia can be caused by neighborhood disorder, social vulnerability, social isolation, and weaker community connections. Environmental factors, including geography and pollution, income, employment, working conditions, socioeconomic status, and poverty may play a role (https://pmc.ncbi.nlm.nih.gov/articles/PMC12301702/).

5. Symptoms
- The most common symptoms of onset Alzheimer's is memory loss that disrupts daily life, difficulty completing daily tasks, and challenges in planning or solving problems (https://www.cdc.gov/alzheimers-dementia/signs-symptoms/alzheimers.html).

6. Diagnosis
- Diagnosis includes many steps:
    - Review of medical and family history 
    - Physical and neurological exams: tests reflexes, muscle tone and strength, sight and hearing, coordination, balance, ability to get up from a chair and walk across the room 
- Lab test 
    - Blood tests can help rule out other potential causes of memory loss, confusion, and other symptoms
    - Newer blood tests are able to measure levels of beta-amyloid protein and tau protein 
- Mental status and neuropsychological testing
    - Just a status check to check memory and thinking skills
- Brain imaging
    - Magnetic resonance imaging (MRI): use radio waves and a strong magnetic field to produce detailed images of the brain 
    - Computerized tomography (CT): specialized X-ray that produces cross sectional images of the brain 
    - PET scan: captures images of the disease process 
        - Fluorodeoxyglucose, also called FDG, PET imaging
        - Amyloid PET imaging
        - Tau PET imaging  (https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/diagnosis-treatment/drc-20350453)

7. Standard of care treatments (& reimbursement)
- There are many types of care needed for patients with Alzheimer's disease as they are progressing. 
    - In Home Care: Home health aides, nurses, or other caregivers may help with daily activities or medical needs.
    - Adult Day Centers: Gives patients the opportunity to be social and participate in fun activities while also giving caregivers well needed breaks. 
    - Long term care: Communal living environment where patients can be provided with care. Examples are retirement housing, assisted living, or nursing homes. 
    - Hospice care: Provided during the last six months of life where the patient is terminally ill. It is used to help manage painful symptoms during the last stages of Alzheimer’s using comfort (https://www.alz.org/help-support/caregiving/care-options) 


8. Disease progression & prognosis
- Preclinical Phase: when the brain changes through the buildup of proteins (10 to 20 yrs before any outward signs appear)
- Mild Stage: when a person begins to feel mild cognitive impairment (misplacing items, forgetting familiar words, etc)
- Moderate Stage: confusion deepens, memory loss grows, personality changes, people may wander off. Require help with daily tasks such as dressing and bathing (longest stage)
- Severe Stage: physical and mental abilities decline (losing ability to speak, walk, swallow, and control their bodily functions), which requires care (https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/in-depth/alzheimers-stages/art-20048448).

9. Continuum of care providers
- Medical care needs to be continuous for doctors appointments to track the progression of the Alzheimer's and to monitor treatment
- Once Alzheimer's worsens at home care is also needed as people with Alzheimer's are no longer able to take care of themselves
- Expenses also go towards continued care for memory retention
- Patients also need to get end of life care as continued care (https://pmc.ncbi.nlm.nih.gov/articles/PMC6925963/)

10. Biological mechanisms (anatomy, organ physiology, cell & molecular physiology) 
- Alzheimer’s affects the hippocampus, the entorhinal cortex, and other connected structures involved in memory. In Alzheimer’s disease, amyloid-beta (Aβ) plaques form outside neurons, while neurofibrillary tangles made of abnormal tau form inside them. There is also a loss of neurons and synapses which results in a cognitive decline. (https://pmc.ncbi.nlm.nih.gov/articles/PMC5290713/)

11. Clinical Trials/next-gen therapies 
- There are currently 158 new drugs across approximately 192 active clinical trials 
    - Phase 3: has 36 drugs in 54 trials 
    - Phase 2: has 84 drugs in 89 trials 
    - Phase 1: has 45 drugs in 49 trials 
- In these trials there are about 73% disease targeting therapies (DTTs), 18% Cognition enhancing symptom targeted therapies, and 10% drugs targeting neuropsychiatric symptoms (https://www.alz.org/news/2026/alzheimers-disease-drug-development-pipeline-is-growing)

## Data Analysis: 

```python
# Import the Patient class and modules needed to read and analyze the dataset.
from sklearn.linear_model import LinearRegression
import pandas as pd
from Patient_Destiny import Patient
import csv
import statistics
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Create Patient objects from every row in the patient CSV dataset.
from pathlib import Path

csv_file = Path.cwd() / "Metadata and Protein Data for Module 1_Patient.csv"
Patient.all_patients.clear()  # empty the list first so re-running doesn't add duplicates
Patient.instantiate_from_csv(csv_file) # Print the first five Patient objects to verify that the CSV was loaded correctly.
print(Patient.all_patients[:5])
```

```python
# Sort patients based on years of education
Patient.all_patients.sort(key=lambda patient: patient.years_education)

# Print the sorted patients
print("\nPatients sorted by years of education:")
print(Patient.all_patients[:10])

# Filter patients with at least 16 years of education and no dementia
filtered_patients = Patient.filter_patients(16, "No dementia")

print("\nPatients with at least 16 years of education and no dementia:")
print(filtered_patients)
```

```python
# Create two groups based on cognitive status
no_dementia = [patient.years_education for patient in Patient.all_patients
               if patient.cognitive_status == "No dementia"]

dementia = [patient.years_education for patient in Patient.all_patients
            if patient.cognitive_status == "Dementia"]

# Calculate the mean and standard deviation for each group
means = [np.mean(no_dementia), np.mean(dementia)]
standard_deviations = [np.std(no_dementia), np.std(dementia)]

# Perform an independent samples t-test
t_stat, p_val = stats.ttest_ind(no_dementia, dementia, equal_var=False)
# Print the t-test results
print(f't_stat = {t_stat}, p_val = {p_val}')
print()
print()
print("Figure 1. Bar Graph: Mean years of education for patients with and without dementia")

# Create a bar graph comparing mean years of education
groups = ["No dementia", "Dementia"]

fig, ax = plt.subplots()

ax.bar(groups, means, yerr=standard_deviations, capsize=5)

ax.set_xlabel("Cognitive Status")
ax.set_ylabel("Mean Years of Education")
ax.set_title("Mean Years of Education by Cognitive Status")

# Add T-test results to the graph
ax.text(
    0.5, 19.5,
    f"T-test: t = {t_stat:.2f}, p = {p_val:.3f}",
    ha="center",
    fontsize=12
)

plt.show()

# T-test comparing years of education between cognitive status groups
education_no_dementia = [
    patient.years_education
    for patient in Patient.all_patients
    if patient.cognitive_status == "No dementia"
]

education_dementia = [
    patient.years_education
    for patient in Patient.all_patients
    if patient.cognitive_status == "Dementia"
]

# Run the independent samples T-test
t_stat, p_val = stats.ttest_ind(
    education_no_dementia,
    education_dementia,
    equal_var=False # t -test: does not assume equal variance 
)

print("\nT-test results:")
print("T-statistic:", t_stat)
print("P-value:", p_val)
```

Analysis of Figure 1: We compared the average years of education for patients with and without dementia. Patients with dementia had a slightly higher average (about 16.6 years) than patients without dementia (about 15.8 years). However, the t-test gave a p-value of 0.170, so this difference was not statistically significant. Based on this dataset, we did not find any relationship between years of education and dementia.

```python
# Create a scatter plot of years of education and age of dementia diagnosis

dementia_patients = [
    patient
    for patient in Patient.all_patients
    if patient.cognitive_status == "Dementia"
    and patient.age_dementia is not None
]

education = [
    patient.years_education
    for patient in dementia_patients
]

age_diagnosis = [
    patient.age_dementia
    for patient in dementia_patients
]

X = np.array(education).reshape(-1, 1)
y = np.array(age_diagnosis)

model = LinearRegression() # create the regression model
model.fit(X, y) # create the best fit line 

r_value, correlation_p = stats.pearsonr(education, age_diagnosis)

plt.figure()

plt.scatter(
    education,
    age_diagnosis
)

# Sort education values so the regression line draws correctly
sort_indices = np.argsort(education) # sort so the line goes from left to right 
X_sorted = X[sort_indices]

print("Figure 2. Scatter Plot: Years of Education and Age at Dementia Diagnosis")
plt.plot(
    X_sorted,
    model.predict(X_sorted)
)

# Add regression results to the graph
plt.text(
    0.05, 0.95,
    f"r = {r_value:.2f}, p = {correlation_p:.3f}, R² = {model.score(X, y):.3f}",
    transform=plt.gca().transAxes,
    va="top"
) # place the text using the graph corners so that the text is on the graph

plt.xlabel("Years of Education")
plt.ylabel("Age at Dementia Diagnosis")
plt.title("Age at Dementia Diagnosis vs. Years of Education")

plt.show()

print("Regression slope:", model.coef_[0])
print("Regression intercept:", model.intercept_)
print("R-squared:", model.score(X, y))
print("Pearson correlation:", r_value)
print("Correlation p-value:", correlation_p)

```

Analysis of Figure 2: We compared each patient's years of education to the age they were diagnosed with dementia. The trend line goes down a little (slope of about -0.27), but there is a lot of variation and the points are spread out. There is little to no correlation (r = -0.08, p = 0.65). Additionally, the p-value is greater than 0.05, so the result is not statistically significant. R² is also very small (0.006), which also signifies that there is no correlation. Based on this dataset, we did not find any relationship between years of education and age at dementia diagnosis.

```python
# create a bar graph of years of education by sex and cognitive status, with a one-way ANOVA
healthy_female = [patient.years_education for patient in Patient.all_patients
                  if patient.sex == "Female" and patient.cognitive_status == "No dementia"]

healthy_male = [patient.years_education for patient in Patient.all_patients
                if patient.sex == "Male" and patient.cognitive_status == "No dementia"]

diseased_female = [patient.years_education for patient in Patient.all_patients
                   if patient.sex == "Female" and patient.cognitive_status == "Dementia"]

diseased_male = [patient.years_education for patient in Patient.all_patients
                 if patient.sex == "Male" and patient.cognitive_status == "Dementia"]

# Mean and standard deviation for each group
group_names = ["Healthy Female", "Healthy Male", "Diseased Female", "Diseased Male"]
group_means = [np.mean(healthy_female), np.mean(healthy_male),
               np.mean(diseased_female), np.mean(diseased_male)]
group_stds = [np.std(healthy_female), np.std(healthy_male),
              np.std(diseased_female), np.std(diseased_male)]

# One-way ANOVA
f_stat, anova_p = stats.f_oneway(healthy_female, healthy_male, diseased_female, diseased_male)
print("\nANOVA results:")
print("F-statistic:", f_stat)
print("p-value:", anova_p)

plt.figure()
plt.bar(group_names, group_means, yerr=group_stds, capsize=5)
plt.xlabel("Sex and Health Status")
plt.ylabel("Mean Years of Education")
plt.title("Years of Education by Sex and Cognitive Status")
plt.text(0.5, 0.95, f"One-way ANOVA: p = {anova_p:.3f}",
         transform=plt.gca().transAxes, ha="center", va="top")
plt.show()
```

Analysis of Figure 3: We used a one-way ANOVA to compare mean years of education across the four groups (healthy female, healthy male, diseased female, diseased male). The result was not statistically significant (F = 0.99, p = 0.402), because the p-value is greater than 0.05. Based on this dataset, we did not find any relationship between years of education and dementia status or sex.
